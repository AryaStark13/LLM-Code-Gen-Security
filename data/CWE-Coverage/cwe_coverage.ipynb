# CWE Coverage Report

This Notebook analyzes the coverage of CWEs across different datasets that we train on and evaluate against. The goal is to understand how well our training data represents the CWEs present in our evaluation datasets.

## Goals
- Identify the CWEs present in each dataset.
- What percentage of CWEs in the evaluation datasets are also present in the training datasets?
- How does our model perform on CWEs that are present in the training data versus those that are not? (Testing Generalization)
- Analyze overlap between different datasets in terms of CWEs. 
- Most and least common CWEs across datasets.

# CodeGuardPlus

## Split
- Eval -- Only used for evaluation

```python
codeguardplus_cwe_ids = [
    20, 79, 95, 119, 125, 170, 215, 326, 347, 416, 502, 643, 732, 787,
    22, 89, 113, 120, 131, 190, 295, 327, 377, 468, 601, 676, 760, 918,
    78, 94, 117, 121, 134, 193, 312, 329, 400, 476, 611, 730, 776, 943
]
```

```python
# overlap between cweval and secodeplt test set
print("CWEVAL and SecodePLT Test Set Overlap CWE IDs:")
print(
    set(cweval_cwe_ids).intersection(set(secodeplt_test_cwe_ids))
)

# overlap between cweval and secodeplt sft set
print("CWEVAL and SecodePLT SFT Set Overlap CWE IDs:")
print(
    set(cweval_cwe_ids).intersection(set(secodeplt_sft_cwe_ids))
)

# overlap between cweval and secodeplt all
print("CWEVAL and SecodePLT All Overlap CWE IDs:")
print(
    set(cweval_cwe_ids).intersection(set(all_secodeplt_cwe_ids))
)
```

```python
print("Count of CWEs in CWEVAL:", len(set(cweval_cwe_ids)))
print("Count of CWEs in SecodePLT Test Set:", len(set(secodeplt_test_cwe_ids)))
```

```python
# cwes present in cweval but not in secodeplt test set
print("CWEVAL CWE IDs not in SecodePLT Test Set:")
print(
    set(cweval_cwe_ids).difference(set(secodeplt_test_cwe_ids))
)
print("Count:", len(
    set(cweval_cwe_ids).difference(set(secodeplt_test_cwe_ids))
))
```

```python
# cwes present in cweval and also in secodeplt all
print("CWEVAL CWE IDs also in SecodePLT All:")
print(
    set(cweval_cwe_ids).intersection(set(all_secodeplt_cwe_ids))
)
print("Count:", len(
    set(cweval_cwe_ids).intersection(set(all_secodeplt_cwe_ids))
))
print("Percentage of CWEVAL covered by SecodePLT All: {:.2f}%".format(
    len(
        set(cweval_cwe_ids).intersection(set(all_secodeplt_cwe_ids))
    ) / len(set(cweval_cwe_ids)) * 100
))
```
