```python
from datasets import load_dataset, DatasetDict
from datasets import Features, Value, List
```

```python
ref_features = Features({
    'task_id': Value('string'),
    'id': Value('string'),
    'CWE_ID': Value('int64'),
    'y_negative': Value('string'),
    'prompt': List({'content': Value('string'), 'role': Value('string')}),
    'cot_steps': Value('string'),
    'completion': List({'content': Value('string'), 'role': Value('string')})
})
```

```python
# Load the dataset
dataset = load_dataset("ShethArihant/CyberNative-DPO-v1")

# Languages to split
languages = ['python', 'java', 'javascript', 'php', 'ruby', 'go']

# Create new dataset dict with train/test splits
new_dataset = DatasetDict()

for lang in languages:
    if lang in dataset:
        # Split 90/10 train/test
        split_data = dataset[lang].train_test_split(test_size=0.1, seed=42)
        
        # Add to new dataset with renamed splits
        new_dataset[f"{lang}_train"] = split_data['train']
        new_dataset[f"{lang}_test"] = split_data['test']
        
        print(f"✓ {lang}: {len(split_data['train'])} train, {len(split_data['test'])} test")
```

```python
new_dataset
```

```python
for split in new_dataset.keys():
    assert new_dataset[split].features == ref_features

# Push back to HuggingFace
new_dataset.push_to_hub(
    "ShethArihant/CyberNative-DPO-v1",
    commit_message="Split each language into train/test (90/10)"
)

print(f"\n✓ Complete! New splits: {list(new_dataset.keys())}")
```
